```python
# python assignment 1 - if else, for loop, while loop


# Q1 movie ticket
age = int(input("Enter your age: "))

if age < 5:
    print("Your ticket is free")
elif age <= 17:
    print("Your ticket price is Rs. 100")
elif age <= 59:
    print("Your ticket price is Rs. 200")
else:
    print("Your ticket price is Rs. 120")


# Q2 electricity bill
units = int(input("Enter units consumed: "))

if units <= 100:
    bill = units * 3
elif units <= 200:
    bill = 100 * 3 + (units - 100) * 5
else:
    bill = 100 * 3 + 100 * 5 + (units - 200) * 8   # first 200 units fixed = 800

print("Total bill: Rs.", bill)


# Q3 multiplication table
num = int(input("Enter a number: "))

for i in range(1, 11):
    print(num, "x", i, "=", num * i)


# Q4 piggy bank
total = 0
days = 0

while total < 1000:
    saved = int(input("Enter today's savings: "))
    total = total + saved
    days = days + 1

print("Goal reached in", days, "days! Total saved: Rs.", total)


# Q5 atm pin
pin = 4321
attempts = 0

while attempts < 3:
    entered = int(input("Enter PIN: "))
    if entered == pin:
        print("Access granted")
        break
    else:
        print("Wrong PIN")
        attempts = attempts + 1

if attempts == 3:
    print("Card blocked")


# Q6 even odd count
n = int(input("Enter N: "))
even = 0
odd = 0

for i in range(1, n + 1):
    if i % 2 == 0:
        even = even + 1
    else:
        odd = odd + 1

print("Even numbers:", even)
print("Odd numbers:", odd)


# Q7 guessing game
secret = 27
guesses = 0
guess = 0

while guess != secret:
    guess = int(input("Guess the number: "))
    guesses = guesses + 1
    if guess > secret:
        print("Too high")
    elif guess < secret:
        print("Too low")

print("Correct! You took", guesses, "guesses.")


# Q8 grade
marks = int(input("Enter marks: "))

if marks < 0 or marks > 100:
    print("Invalid marks")    # check this first
elif marks >= 90:
    print("Grade: A")
elif marks >= 75:
    print("Grade: B")
elif marks >= 50:
    print("Grade: C")
elif marks >= 35:
    print("Grade: D")
else:
    print("Fail")


# Q9 sum of digits
bill_no = int(input("Enter bill number: "))
total = 0

while bill_no > 0:
    digit = bill_no % 10       # last digit
    total = total + digit
    bill_no = bill_no // 10    # remove last digit

print("Sum of digits:", total)

if total > 20:
    print("Congratulations! You win a free gift.")
else:
    print("Sorry, no gift this time.")


# Q10 staircase
steps = int(input("Enter number of steps: "))

for i in range(1, steps + 1):
    for j in range(i):
        print("*", end=" ")
    print()   # next line
```
