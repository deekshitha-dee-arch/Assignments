```python
# Python Assignment 2 - loops, if else, Lists, Strings
# Q1 class marks report
marks = [78, 45, 92, 33, 67, 88, 51, 29]

passed = 0
for m in marks:
    if m >= 35:
        passed = passed + 1

print("Highest marks:", max(marks))
print("Lowest marks:", min(marks))
print("Class average:", sum(marks) / len(marks))
print("Students passed:", passed)
```

```python
# Q2 grocery shopping list
shopping = []

while True:
    item = input("Enter item: ")
    if item == "done":
        break
    if item in shopping:
        print("Already added")
    else:
        shopping.append(item)

count = 1
for item in shopping:
    print(str(count) + ".", item)
    count = count + 1

print("Total items:", len(shopping))
```

```python
# Q3 username validator
username = input("Enter username: ")

if len(username) < 5 or len(username) > 12:
    print("Invalid: Username must be 5 to 12 characters")
elif " " in username:
    print("Invalid: Username cannot have spaces")
elif username[0].isdigit():
    print("Invalid: Username cannot start with a digit")
else:
    print("Valid username")
```

```python
# Q4 vowel counter
tweet = input("Enter a sentence: ")
vowels = 0
consonants = 0
spaces = 0

for ch in tweet.lower():
    if ch in "aeiou":
        vowels = vowels + 1
    elif ch.isalpha():          # letter but not vowel
        consonants = consonants + 1
    elif ch == " ":
        spaces = spaces + 1

print("Vowels:", vowels)
print("Consonants:", consonants)
print("Spaces:", spaces)
```

```python
# Q5 palindrome name
name = input("Enter your name: ")
name_lower = name.lower()       # so Anna and anna both work

if name_lower == name_lower[::-1]:
    print("Bonus! Your name is a palindrome.")
else:
    print("Sorry, not a palindrome.")
```

```python
# Q6 temperature tracker
temps = [32, 35, 38, 41, 36, 29, 40]
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

print("Heat alert days:")
for i in range(len(temps)):
    if temps[i] > 37:
        print(days[i], "-", temps[i])

hottest = max(temps)
hottest_day = days[temps.index(hottest)]
print("Hottest day:", hottest_day, "with", hottest)

fahrenheit = []
for t in temps:
    fahrenheit.append(t * 9 / 5 + 32)

print("In Fahrenheit:", fahrenheit)
```

```python
# Q7 words in a sentence
sentence = input("Enter a sentence: ")
words = sentence.split()

longest = words[0]
shortest = words[0]
capitals = 0

for w in words:
    if len(w) > len(longest):
        longest = w
    if len(w) < len(shortest):
        shortest = w
    if w[0].isupper():
        capitals = capitals + 1

print("Longest word:", longest)
print("Shortest word:", shortest)
print("Words starting with capital letter:", capitals)
```

```python
# Q8 negative balances
balances = [1500, -200, 3000, -50, 0, 750, -1000]
positive_accounts = []
negative_accounts = []

for b in balances:
    if b > 0:
        positive_accounts.append(b)
    elif b < 0:
        negative_accounts.append(b)
    # 0 goes in neither

print("Positive accounts:", positive_accounts)
print("Negative accounts:", negative_accounts)
print("Total owed:", -sum(negative_accounts))
```

```python
# Q9 password strength
password = input("Enter password: ")
has_upper = False
has_lower = False
has_digit = False

for ch in password:
    if ch.isupper():
        has_upper = True
    elif ch.islower():
        has_lower = True
    elif ch.isdigit():
        has_digit = True

score = 0
if has_upper:
    score = score + 1
if has_lower:
    score = score + 1
if has_digit:
    score = score + 1
if len(password) >= 8:
    score = score + 1

if score == 4:
    print("Strong")
elif score >= 2:
    print("Medium")
else:
    print("Weak")
```

```python
# Q10 initials generator
full_name = input("Enter full name: ")
parts = full_name.split()

initials = ""
for p in parts:
    initials = initials + p[0].upper() + "."
print("Initials:", initials)

print("Title case:", full_name.title())

short_name = ""
for p in parts[:-1]:            # all except surname
    short_name = short_name + p[0].upper() + ". "
short_name = short_name + parts[-1].title()
print("Short name:", short_name)
```
