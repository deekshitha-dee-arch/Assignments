```python
# Python Assignment 3 - Loops, if-else, Sets, Tuples, Dictionaries
# Q1 phone contact book
contacts = {}

while True:
    print("1. Add contact")
    print("2. Search contact")
    print("3. Show all contacts")
    print("4. Exit")
    choice = input("Enter choice: ")

    if choice == "1":
        name = input("Enter name: ")
        if name in contacts:
            print("Contact already exists")
        else:
            phone = input("Enter phone number: ")
            contacts[name] = phone
            print("Contact added")
    elif choice == "2":
        name = input("Enter name to search: ")
        if name in contacts:
            print(name, ":", contacts[name])
        else:
            print("Contact not found")
    elif choice == "3":
        if len(contacts) == 0:
            print("No contacts yet")
        for name, phone in contacts.items():
            print(name, ":", phone)
    elif choice == "4":
        print("Bye")
        break
    else:
        print("Invalid choice")
```

```python
# Q3 student marks dictionary
marks = {"Aman": 82, "Priya": 91, "Rahul": 45, "Sneha": 67, "Kiran": 30}

for name, m in marks.items():
    if m >= 40:
        print(name, "- Pass")
    else:
        print(name, "- Fail")

topper = ""
top_marks = 0
for name, m in marks.items():
    if m > top_marks:
        top_marks = m
        topper = name
print("Topper:", topper, "with", top_marks)

passed = {}
for name, m in marks.items():
    if m >= 40:
        passed[name] = m
print("Passed students:", passed)
```

```python
# Q4 common friends
riya_friends = {"Aman", "Neha", "Karan", "Sita", "Vikram"}
arjun_friends = {"Neha", "Vikram", "Pooja", "Aman", "Rohan"}

print("Mutual friends:", riya_friends & arjun_friends)       # intersection
print("Only Riya's friends:", riya_friends - arjun_friends)  # difference
print("Everyone:", riya_friends | arjun_friends)             # union
print("Suggestions for Riya:", arjun_friends - riya_friends)
```

```python
# Q5 character counter
text = input("Enter text: ")
counts = {}

for ch in text:
    if ch == " ":
        continue
    if ch in counts:
        counts[ch] = counts[ch] + 1
    else:
        counts[ch] = 1

print(counts)

most = ""
most_count = 0
for ch, c in counts.items():
    if c > most_count:
        most_count = c
        most = ch
print("Most frequent character:", most, "(" + str(most_count) + " times)")
```

```python
# Q6 restaurant billing
menu = {"dosa": 60, "idli": 40, "vada": 30, "coffee": 25, "tea": 15}
total = 0

while True:
    item = input("What would you like to order: ").lower()
    if item == "bill":
        break
    if item in menu:
        total = total + menu[item]
        print(item, "added - Rs.", menu[item])
    else:
        print("Item not available")

print("Total: Rs.", total)
if total > 200:
    discount = total * 10 / 100
    total = total - discount
    print("10% discount applied: Rs.", discount)
print("Final bill: Rs.", total)
```

```python
# Q7 city coordinates
cities = [("Hyderabad", 17.38, 78.48),
          ("Mumbai", 19.07, 72.87),
          ("Chennai", 13.08, 80.27),
          ("Delhi", 28.61, 77.20)]

for city, lat, lon in cities:      # unpacking tuple
    print(city, "- Latitude:", lat, "Longitude:", lon)

north_city = ""
max_lat = 0
for city, lat, lon in cities:
    if lat > max_lat:
        max_lat = lat
        north_city = city
print("Farthest north:", north_city, "(" + str(max_lat) + ")")

# cities[1][1] = 20.0
# TypeError: 'tuple' object does not support item assignment
# tuples are immutable, values cant be changed after creating
# to change it we have to make a new tuple
cities[1] = ("Mumbai", 20.0, 72.87)
print(cities[1])
```

```python
# Q8 library book tracker
books = {"Python Basics": 3, "Data Science 101": 0, "Machine Learning": 2, "SQL Guide": 1}

book = input("Enter book name: ")

if book not in books:
    print("Book not found")
elif books[book] == 0:
    print("Currently unavailable")
else:
    books[book] = books[book] - 1
    print("Book issued successfully")

print(books)
```

```python
# Q9 group employees by department
employees = [("Ravi", "IT"), ("Anjali", "HR"), ("Suresh", "IT"),
             ("Meena", "Finance"), ("Arun", "HR"), ("Divya", "IT")]

dept = {}
for name, d in employees:
    if d not in dept:
        dept[d] = []        # new department, empty list
    dept[d].append(name)

print(dept)

for d, names in dept.items():
    print(d, ":", len(names), "employees")
```

```python
# Q10 quiz checker
answer_key = ("B", "C", "A", "D", "B")
wrong = set()
score = 0

for i in range(5):
    ans = input("Answer for Q" + str(i + 1) + ": ").upper()
    if ans == answer_key[i]:
        print("Q" + str(i + 1), "correct")
        score = score + 1
    else:
        print("Q" + str(i + 1), "wrong")
        wrong.add(i + 1)

print("Wrong questions:", wrong)
print("Score:", score, "/ 5")

if score == 5:
    print("Excellent")
elif score >= 3:
    print("Good")
else:
    print("Needs improvement")
```
